Matplotlib is an open-source Library for creating Static & interactive visualization in python

MatplotLib pyplot - Pyplot is a Module within MatplotLib that a MATLAB like Interface for Making Plots.
                  - It Simplifies the Process of Adding plot elements such as Lines, Images & Text to the axes of Current Figure.

KEY ADVANTAGES OF MATPLOTLIB:
1. It Supports Line Plots, Bar Charts, Histograms, Scatter Plots & 3D Visualization.
2. It's compatible with Multiple General Purpose GUI Toolkit
3. It is Capable of Generating High Quality Outputin Various Hardcopy Formats
4. Seamless Integration with Other python Libraries

```python
# Importing the required libraries
import matplotlib.pyplot as  plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)
z = np.cos(x)
print(x)
print(y)
print(z)
```

PLOTTING THE BASIC GRAPH

```python
# Plotting Sine Wave
plt.plot(x,y)
plt.show()
```

```python
# Plotting Cosine Wave
plt.plot(x,z)
plt.show()
```

```python
# Adding x-axis & y-axis labels
plt.plot(x,y)
plt.xlabel('angles')
plt.ylabel('sine value')
plt.title('sine wave')
plt.show()
```

```python
# Parabola
x =np.linspace(-10,10,100)
y = x**2
plt.plot(x,y)
plt.title('Parabola')
plt.show()
```

```python
# Altering the color and style of the plot
plt.plot(x,y, 'r+')
plt.show()
```

```python
# Plotting  Multiple Plots in a Single Graph
x = np.linspace(-5, 5, 100)
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x))
plt.show()
```

PIE CHART

```python
fig2 = plt.figure()
ax = fig2.add_axes([0,0,1,1])
languages = ['c++','python', 'ruby', 'java', 'SQL']
people = [76,39, 89,67,40]
ax.pie(people , labels = languages, autopct = '%1.1f%%')
plt.show()
```

SCATTER PLOT

```python
x = np.linspace(0,10,30)
y = np.sin(x)
z = np.cos(y)
fig2 = plt.figure()
ax = fig2.add_axes([0,0,1,1])
ax.scatter(x,y,color = 'r')
ax.scatter(x,z, color = 'b')
plt.show()
```

3-D SCATTER PLOT

```python
fig3 = plt.figure()
ax = fig3.add_axes([0,0,1,1])
z = 20 * np.random.random(1000)
x = np.sin(z)
y = np.cos(z)
ax.scatter(x,y,z ,c = z, cmap ='Blues')
plt.show()
```
