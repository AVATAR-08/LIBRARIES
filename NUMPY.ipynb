Numpy Arrays

```python
# CREATING NUMPY ARRAY
A1 = np.array([1,2,3,4,5])                       # 1D ARRAY
A2 = np.array([(1,2,3,4) , (5,6,7,8)])           # 2D ARRAY
A3 = np.array([[[1,2],[3,4]] , [[5,6],[7,8]]])   # 3D ARRAY
print(A1) 
print(A2)   
print(A3)   
```

INITIAL PLACEHOLDERS IN NUMPY ARRAYS

```python
#create a numpy array of zeros
x  = np.zeros((4,5))
print(x)
```

```python
#creates a numpy array of oness
y = np.ones((4,4))
print(y)
```

```python
#create a identity matrix
a = np.eye(5)
print(a)
```

```python
#create a  numpy array  with ramdom values

b = np.random.random((3,3))
print(b)
```

```python
#random integer values array  with apecific range
c = np.random.randint(10,100, (3,3))
print(c)
```

```python
#array of evenly spaced values -> specifying the number of values reuired
d = np.linspace(10,30,6)
print(d)
```

```python
#array of evenly spaced values -> specifyiing the step
e = np.arange(10,30,5)
print(e)
```

MATHEMATICAL OPERATIOMNS ON NUMPY ARRAY

```python
a = np.random.randint(10, 100, (3,3))
b = np.random.randint(10, 100, (3,3))

print(a)
print(b)
```

```python
# ARITHMETIC OPERATIONS ON NUMPY ARRAYS
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
# BINARY OPERATIONS ON NUMPY ARRAYS
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
```

ARRAY MANIPULATION

```python
array =np.random.randint(20,200, (4,5))
print(array)
print(array.shape)
```

```python
#tranpose
trans = np.transpose(array)
print(trans)
print(trans.shape)
```
